## Advanced: Running the Heuristic

> ⭐ **Optional section** — feel free to skip on first pass.

The next cell scores two different team shapes. You should see a strong BUILD recommendation for a team like this repo's own sandbox, and a strong BUY recommendation for a team rolling a simple policy out across many repos.

```python
from labs.lab_22_buy_vs_build import BuyBuildFactors, recommend

like_this_repo = BuyBuildFactors(
    needs_custom_risk_model=True,
    rolling_out_to_many_repos=False,
    has_in_house_actions_expertise=True,
    budget_for_a_paid_tool=False,
)
result = recommend(like_this_repo)
print(
    f"recommendation: {result['recommendation'].upper()} (build={result['build_score']}, buy={result['buy_score']})"
)
for reason in result["reasons"]:
    print(f"  - {reason}")

org_wide = BuyBuildFactors(
    needs_custom_risk_model=False,
    rolling_out_to_many_repos=True,
    has_in_house_actions_expertise=False,
    budget_for_a_paid_tool=True,
)
result2 = recommend(org_wide)
print(
    f"\nrecommendation: {result2['recommendation'].upper()} (build={result2['build_score']}, buy={result2['buy_score']})"
)
for reason in result2["reasons"]:
    print(f"  - {reason}")
```
