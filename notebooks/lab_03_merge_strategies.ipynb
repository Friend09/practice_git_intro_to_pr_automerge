## 2. Live Mode: Build a Real Disposable Repo

The next cell only does something real in `PRA_MODE=live`: it builds an actual repo with `build_demo_repo`, then applies all three strategies to independent copies. In fixture mode it just explains what would happen -- re-run with `PRA_MODE=live` to see real `git log` output instead of the canned diagrams above.

```python
if PRA_MODE == "live":
    import shutil
    import tempfile
    from labs.lab_03_merge_strategies import (
        build_demo_repo,
        apply_merge_commit,
        apply_squash,
        apply_rebase,
    )

    with tempfile.TemporaryDirectory(prefix="pra_lab03_nb_") as tmp:
        base = build_demo_repo(Path(tmp))
        for name, fn in [
            ("merge_commit", apply_merge_commit),
            ("squash", apply_squash),
            ("rebase", apply_rebase),
        ]:
            copy = Path(tmp) / f"copy_{name}"
            shutil.copytree(base, copy)
            print(f"--- {name} (live) ---")
            print(fn(copy))
else:
    print("Fixture mode: re-run with PRA_MODE=live to build a real disposable repo")
    print("and see genuine git log --graph output for all three strategies.")
```
